### React Agent

- start -> reasoning either go to tool call or end
- configure the LangGraph react agent with model + tools + state + store for long term memory + static schema for static configurations
- only in react agent in langGraph, a tool can return a command if you want to update messages + state and it would be returned as command without wrapping it in tool message
- create_react_agent(
    model, prompt=function to get system message, state_schema
)
