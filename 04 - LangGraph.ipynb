### Agentic Patterns

1. Reflection Pattern:
   - LLm output -> another LLm reflect/observe to validate or observe the output of the first llm output and return it to the llm to further update it or return it to the user
   - user to assess the quality of the output of the LLM 
   - planner -> llm -> go either to the planner or return it to the user

2. Tool Use Patter:
   - model is enabled to use the tool
   - "ToolNode(config with tools)" takes ai message as an arg of type tool call and it will execute the tool and return tool message back to the LLM
   - LLMbined with tools-> output ai message-> conditional edge wether ai message has content or tool call forwards it to node or tool node
